```python
import pandas as pd

buyer = pd.read_csv("tickit_public_buyer.txt", sep="|")
seller = pd.read_csv("tickit_public_seller.txt", sep="|")
venue = pd.read_csv("tickit_public_venue.txt", sep="|")
category = pd.read_csv("tickit_public_category.txt", sep="|")
event = pd.read_csv("tickit_public_event.txt", sep="|")
listing = pd.read_csv("tickit_public_listing.txt", sep="|")
sale = pd.read_csv("tickit_public_sale.txt", sep="|")
```

```python
tables = {
    "buyer": buyer,
    "seller": seller,
    "venue": venue,
    "category": category,
    "event": event,
    "listing": listing,
    "sale": sale
}

for name, df in tables.items():
    print(f"--- {name} ---")
    print(df.shape)
    print(df.head(2))
    print()
```
